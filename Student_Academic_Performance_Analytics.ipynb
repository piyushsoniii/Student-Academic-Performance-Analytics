## Methodology and decisions
- Missing `Attendance_Percent` values are replaced with the column median. The median is less sensitive to extreme observations than the mean.
- Duplicate Student_ID records are removed if present (first occurrence retained).
- Subject marks are checked against the inclusive 0–100 range. Invalid score rows are excluded from score-based analysis and are not silently modified.
- Average marks are the arithmetic mean of Python, Maths and Physics marks.
- Grades: A+ (90–100), A (80–89.99), B (70–79.99), C (60–69.99), D (50–59.99), E (40–49.99), F (below 40).
- At-risk rule: average marks below 40%.
- Attendance bands: below 60%, 60–74.99%, 75–89.99%, and 90–100%.

## Interpretation and limitations
In this dataset, the mean average marks are close across departments. The observed correlation between daily study hours and average marks is very weak; this does not establish that study time has no effect, nor does it prove causation. The internet-access groups also have similar mean marks. The analysis is descriptive and is limited to the fields and records supplied in this synthetic training dataset.

## Academic-support recommendations
1. Review attendance and assessment progress periodically, especially for students in the lowest attendance band.
2. Offer targeted subject support based on individual subject scores rather than relying only on an overall average.
3. Provide structured study-planning guidance and monitor progress over several assessments.
4. Ensure students without reliable internet access can use campus labs, downloadable material or offline resources.
5. Recheck students flagged by low marks or unusual score patterns with faculty before assigning support; an automated flag should prompt review, not a final judgement.
